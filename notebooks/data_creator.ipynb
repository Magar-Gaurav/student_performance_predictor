**This python notebook creates dataset which i will be working on**

```python
# importing numpy
import numpy as np
```

```python
# for reproducible results
np.random.seed(42)

# defining the number of dataset i would need (i just needed 200 or less than that so i used 200 as n)
n = 200

StudyHours = np.random.randint(1,11,n)
Attendence = np.random.randint(60,101,n)
Assignments = np.random.randint(0,11,n)

# Generating Marks on the basis of other features
noise = np.random.normal(0,4,n)

Marks = (
    4 * StudyHours 
    + 0.15 * Attendence
    + 1.5 * Assignments
    + 35
    + noise
)

# Keeping marks between 0 and 100
Marks = np.clip(Marks, 0, 100).round().astype(int)

# Combining all the columns into one dataset
dataset = np.column_stack((
    StudyHours,
    Attendence,
    Assignments,
    Marks
))

# Saving the dataset as a CSV file
np.savetxt(
    r"C:\ai_ml\day14\student_performance_predictor\data\student_dataset.csv",
    dataset,
    delimiter = ",",
    header = "StudyHours,Attendence,Assignments,Marks",
    comments="",
    fmt = "%d"    
)

print("Dataset created successfully!!!")
print("Dataset shape:", dataset.shape)
print("\nFirst 5 rows:\n",dataset[:5])
```
